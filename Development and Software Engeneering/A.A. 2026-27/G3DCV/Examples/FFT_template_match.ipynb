```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Template matching in frequency domain

```python
!wget -nc https://www.dais.unive.it/~bergamasco/teachingfiles/cvstuff/murales1.png
!wget -nc https://www.dais.unive.it/~bergamasco/teachingfiles/cvstuff/murales2.png
```

```python
I1 = cv.imread("murales1.png", cv.IMREAD_GRAYSCALE).astype( np.float32 )
I2 = cv.imread("murales2.png", cv.IMREAD_GRAYSCALE).astype( np.float32 )
```

```python
plt.figure( figsize=(12,12))
plt.subplot(1,2,1)
plt.imshow(I1, cmap="gray")
plt.subplot(1,2,2)
plt.imshow(I2, cmap="gray")
```

```python
# Convert to frequency domain
I1f = np.fft.fftshift( np.fft.fft2( I1 ) )
I2f = np.fft.fftshift( np.fft.fft2( I2 ) )

plt.figure( figsize=(12,12))
plt.subplot(1,2,1)
plt.imshow( np.log( np.abs(I1f)+1 ) )
plt.subplot(1,2,2)
plt.imshow( np.log( np.abs(I2f)+1 ) )
```

Correlation between I1 and I2 can be computed by multiplying I1f with the complex conjugate of I2f

```python
I1convI2 = np.real( np.fft.ifft2( np.fft.ifftshift( I1f * np.conj(I2f) ) ) )
plt.imshow(I1convI2, cmap="gray")
```

```python
t_row,t_col = np.unravel_index( np.argmax(I1convI2), I1convI2.shape )
print(t_row, " ", t_col)
```

```python
I2t = cv.warpAffine( I2, np.array([ [1,0,t_col], [0,1,t_row] ], dtype=np.float32), dsize=I2.shape  )
plt.figure()
plt.imshow(I2t, cmap="gray")
plt.figure()
plt.imshow(I1, cmap="gray")

cv.imwrite("a.jpg", I1)
cv.imwrite("b.jpg", I2t)
```
