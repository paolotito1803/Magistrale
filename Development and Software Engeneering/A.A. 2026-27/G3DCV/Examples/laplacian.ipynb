```python
I = cv.imread("moon.png", cv.IMREAD_GRAYSCALE)
I = I.astype(np.float32)
```

[Laplacian filtering documentation](https://docs.opencv.org/4.6.0/d4/d86/group__imgproc__filter.html#gad78703e4c8fe703d479c1860d76429e6)

```python
from ipywidgets import interact, widgets

def laplacian( boost ):
    plt.figure( figsize=(20,80))
    Ilap = cv.Laplacian(I, cv.CV_32F, ksize=3 )
    Ifilt = np.clip( I-Ilap*boost, 0, 255 )

    plt.subplot(1,3,1)
    plt.imshow( I, interpolation="none", vmin=0, vmax=255, cmap="gray" );
    plt.title("original")
    plt.subplot(1,3,3)
    plt.imshow( Ilap*boost, interpolation="none", cmap="gray", vmin=-100, vmax=100 );
    plt.title("Laplacian*boost")
    plt.subplot(1,3,2)
    plt.imshow( Ifilt, interpolation="none", cmap="gray", vmin=0, vmax=255 );
    plt.title("Filtered")
    plt.show()

interact( laplacian, boost=(0,1.0,0.025))
```
