```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets

!wget https://www.dais.unive.it/~bergamasco/teachingfiles/cvstuff/chessboard.png
```

```python
I = cv.imread("chessboard.png")

K = np.array( [ [I.shape[1], 0, I.shape[1]/2],  [0, I.shape[0], I.shape[0]/2], [0,0,1] ]  )
print(K)
```

```python

def undistort_and_plot( k1=0,k2=0,k3=0,p1=0,p2=0 ):

    k = np.array( [k1,k2,p1,p2,k3] )
    Iun = cv.undistort( I, K, k )

    plt.figure( figsize=(6,6))
    plt.imshow( Iun );

interact( undistort_and_plot, k1=(-1,1,0.01), k2=(-1,1,0.01), k3=(-1,1,0.01), p1=(-1,1,0.01), p2=(-1,1,0.01) )
```
