```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
import scipy
```

# Image deblurring demo

```python
!wget -nc https://www.dais.unive.it/~bergamasco/teachingfiles/cvstuff/traffic.png
```

```python
I = cv.imread( "traffic.png", cv.IMREAD_GRAYSCALE )

G = np.fft.fft2( I )

plt.figure( figsize=(10,10) )
plt.imshow(  I , cmap="gray")

plt.figure( figsize=(10,20) )
plt.subplot(1,2,1 )
plt.imshow(  I , cmap="gray")
plt.title("Image")
plt.subplot(1,2,2 )
plt.imshow( np.log( np.fft.ifftshift(np.abs(G))), interpolation="none")
plt.title("Power spectrum")
#plt.imshow( np.log( (np.abs(G))))
```

```python
# Mask the image boundaries to reduce artefacts

w = np.expand_dims( scipy.signal.windows.tukey( I.shape[0], 0.15) , axis=-1 )
plt.figure()
plt.plot(w)
plt.grid()

I = I * (w@w.T)
G = np.fft.fft2( I )

plt.figure( figsize=(20,10) )
plt.subplot(1,2,1 )
plt.imshow(  I , cmap="gray")
plt.subplot(1,2,2 )
plt.imshow( np.log( np.fft.ifftshift(np.abs(G))))
```

```python
w = np.expand_dims( scipy.signal.windows.gaussian( 7, 3.0 ), axis=-1)

PSF = w @ w.T

PSF = PSF / np.sum(PSF)

plt.figure( figsize=(3,3) )
plt.imshow( PSF, interpolation=None)
```

```python
G = np.fft.fft2( I )
H = np.fft.fft2( PSF, I.shape )

plt.figure( figsize=(20,10) )
plt.subplot(1,2,1 )
plt.imshow(  np.fft.ifftshift(np.abs(H)))
plt.title("H power spectrum")
plt.colorbar()
plt.subplot(1,2,2 )
plt.imshow( np.log( np.fft.ifftshift(np.abs(G))))
plt.title("G power spectrum")
plt.colorbar()

noise_to_signal_ratio = 0.0001
Fhat = np.conj(H) / ( (np.conj(H)*H) + noise_to_signal_ratio ) * G

plt.figure( figsize=(10,10) )
plt.imshow( np.log( np.fft.ifftshift(np.abs(Fhat))))
plt.title("Deblurred image power spectrum")
plt.colorbar()
```

```python
Ideb = np.real( np.fft.ifft2(Fhat) )

plt.figure( figsize=(20,10) )
plt.subplot(1,2,1)
plt.imshow(I, cmap="gray")
plt.title("Original image")
plt.subplot(1,2,2)
plt.title("Deblurred image")
plt.imshow(Ideb, cmap="gray")
```
